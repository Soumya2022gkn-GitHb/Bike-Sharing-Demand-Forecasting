# Bike-Sharing Demand Forecasting

This project is organized as a clean, production-minded Python workflow for Part 1 of the bike-sharing demand forecasting task. Run the notebooks in order from `notebooks/1_data_ingestion.ipynb` through `notebooks/6_Building_Streamlit_UI.ipynb`.

## Folder Structure

```text
Bike-Sharing-Demand-Forecasting/
|-- app/
|   `-- streamlit_app.py
|-- data/
|   |-- raw/
|   `-- processed/
|-- models/
|-- notebooks/
|   |-- 1_data_ingestion.ipynb
|   |-- 2_feature_engineering.ipynb
|   |-- 3_training.ipynb
|   |-- 4_evaluation.ipynb
|   |-- 5_plot_generation.ipynb
|   `-- 6_Building_Streamlit_UI.ipynb
|-- reports/
|   `-- figures/
|-- src/
|   `-- bsdf/
`-- tests/
```


## Notebook Order

1. `1_data_ingestion.ipynb`: load data, validate data, and preprocess data.
2. `2_feature_engineering.ipynb`: add time features, encode categorical variables, and scale numeric variables.
3. `3_training.ipynb`: train and compare Linear Regression, Random Forest, Gradient Boosting, and XGBoost.
4. `4_evaluation.ipynb`: compare models, analyze errors, and produce forecast analysis.
5. `5_plot_generation.ipynb`: generate demand trend, feature importance, error distribution, heat map, weather demand, and model comparison plots.
6. `6_Building_Streamlit_UI.ipynb`: document and validate the Streamlit decision dashboard.

The selected business model is chosen from the full model comparison by the lowest mean absolute deviation. In the executed workflow, XGBoost is best with a mean absolute deviation of 50.91 hourly rentals.
