# 6. Building Streamlit UI

This notebook documents and validates the interactive Streamlit dashboard for the Bike-Sharing Demand Forecasting project. The dashboard lets a business user change input parameters and immediately see how the forecast changes across Linear Regression, Random Forest, Gradient Boosting, and XGBoost.

````python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT / "src"))

from bsdf.config import get_config
from bsdf.plotting import MODEL_DISPLAY_NAMES, SEASON_LABELS, WEATHER_LABELS
from bsdf.scenario import build_scenario_features, predict_scenario
from bsdf.utils import load_model

CONFIG = get_config(PROJECT_ROOT)
APP_PATH = PROJECT_ROOT / "app" / "streamlit_app.py"
````

## UI Position In The Pipeline

```text

```

The app reads trained artifacts from `models/` and processed comparison files from `data/processed/`. It does not retrain models inside the web app.

## Interactive Dashboard Controls

The Streamlit app gives users direct controls for:

- Model selection: Linear Regression, Random Forest, Gradient Boosting, XGBoost.
- Calendar inputs: season, month, hour, weekday, working day, holiday, historical year pattern.
- Weather inputs: weather situation, temperature, feels-like temperature, humidity, and windspeed.

When any input changes, Streamlit reruns the app and refreshes the forecast immediately. The dashboard also shows a side-by-side prediction table and bar chart for all available models.

````python
required_artifacts = {
    "Streamlit app": APP_PATH,
    "Model comparison": CONFIG.processed_dir / "model_comparison.csv",
    "Linear Regression model": CONFIG.model_dir / "linear_regression.joblib",
    "Random Forest model": CONFIG.model_dir / "random_forest.joblib",
    "Gradient Boosting model": CONFIG.model_dir / "gradient_boosting.joblib",
    "XGBoost model": CONFIG.model_dir / "xgboost.joblib",
}

artifact_status = pd.DataFrame(
    [
        {"artifact": name, "path": str(path), "exists": path.exists()}
        for name, path in required_artifacts.items()
    ]
)
artifact_status
````

````python
model_files = {
    "linear_regression": CONFIG.model_dir / "linear_regression.joblib",
    "random_forest": CONFIG.model_dir / "random_forest.joblib",
    "gradient_boosting": CONFIG.model_dir / "gradient_boosting.joblib",
    "xgboost": CONFIG.model_dir / "xgboost.joblib",
}
models = {name: load_model(path) for name, path in model_files.items() if path.exists()}

scenario_features = build_scenario_features(
    season=3,
    year=1,
    month=9,
    hour=17,
    holiday=0,
    weekday=5,
    workingday=1,
    weather=1,
    temperature_celsius=24.0,
    feels_like_celsius=27.0,
    humidity_percent=55.0,
    windspeed_normalized=0.20,
)

scenario_predictions = predict_scenario(models, scenario_features)
scenario_predictions
````

## Example Scenario Meaning

The validation cell above simulates a business user selecting a fall working-day evening peak hour with clear weather. In the web app, changing any of those values updates the predicted hourly rental count and the model comparison table.

````python
pd.DataFrame(
    {
        "season_code": list(SEASON_LABELS.keys()),
        "season_label": list(SEASON_LABELS.values()),
        "weather_code": list(WEATHER_LABELS.keys()),
        "weather_label": list(WEATHER_LABELS.values()),
    }
)
````

## Run The Interactive Dashboard

From the project root, run:

```bash
streamlit run app/streamlit_app.py
```

The dashboard is intended for decision review. It keeps XGBoost as the best model from the offline comparison, but lets users inspect how different trained models react to the same operational scenario.
